## 7 | Executive Summary & Key Findings

### Project Overview

This project uses SQL and Python to analyze a simulated 2026 FIFA World Cup dataset, focusing on Spain and Argentina.

The objective is to compare team performance, identify individual player contributions, and explore the factors associated with tournament success.


### Data Source & Methodology

**Dataset:** [World Cup 2026 Open Match Data](https://github.com/Alamyy/Worldcup26)

**Dataset creator:** Hesham El Alamy / Alamy AI

**Underlying source:** FIFA Training Centre post-match reports

**License:** Creative Commons Attribution 4.0 (CC BY 4.0)

The dataset contains structured match, team, and player-level information covering 104 matches and 48 teams.

For this project, I imported the data into SQLite, validated relationships across tables, and used SQL to analyze team performance, individual contributions, and scoring patterns.

Python, pandas, and Matplotlib were used to present the results through visualizations.

**Note:** This portfolio analysis treats the dataset as a simulated tournament scenario and does not present its results as verified real-world outcomes.


### Key Findings

**1. Argentina demonstrated stronger attacking performance.**

Argentina scored 19 goals across eight matches (2.38 per match), compared with Spain's 14 goals (1.75 per match).

**2. Spain demonstrated superior defensive consistency.**

Spain conceded only one goal throughout the tournament and maintained a clean sheet in seven of eight matches (87.5%). Argentina conceded eight goals and recorded two clean sheets (25%).

**3. Argentina relied more heavily on its leading scorer.**

Lionel Messi contributed eight goals, representing 42.1% of Argentina's total. Mikel Oyarzabal led Spain with five goals, representing 35.7% of the team's total.

**4. Goal-event data required reconciliation.**

Player goal events initially differed from match-level totals. SQL reconciliation identified one opponent own goal benefiting each team, resolving both discrepancies.

### Conclusion

In this simulated tournament, Spain defeated Argentina despite scoring fewer goals overall.

The results suggest that defensive consistency was an important distinguishing factor in Spain's success. Argentina's greater attacking output did not translate into a tournament victory.
